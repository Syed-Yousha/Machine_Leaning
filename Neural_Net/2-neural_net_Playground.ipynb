```python
df['date'] = pd.to_datetime(df['date']) #converted date into date time object
```

```python
df['date']
```

```python
df['year'] = df['date'].apply(lambda date: date.year) #Making a new column for year
```

```python
#the lambda expression is equavelent to a function
# our lambda expression is equals to:

# def year_ext(dare):
#     return date.year
```

```python
df['month'] = df['date'].apply(lambda date: date.month) #New col for month too
```

```python
df.head()
```

```python
#lets check the corr of month with price
plt.figure(figsize=(10,6))
sns.boxplot(x='month', y='price', data=df)
```
