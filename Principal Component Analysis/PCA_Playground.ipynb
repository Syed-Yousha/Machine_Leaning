```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer() #load method is used for sklearn builtin datasets
```

```python
type(cancer) # Datasets and DataFrames are not same 
```

```python
cancer.keys() #Dataset column names
```

```python
print(cancer['DESCR'])
```

```python
#Making cancer dataframe
df = pd.DataFrame(cancer['data'], columns=cancer['feature_names'])
df.head()
```

```python
cancer['target'] #to check the target variable
```

```python
cancer['target_names'] #names of target, this col is really helpful when we are predicting from ML models
```

```python
#making all the columns of same scale
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

```python
# PCA: pca is used for when we are reducing variance and rows of a datas, use to improve accuracy of linear data.
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca.fit(scaled_data)
```

```python
x_pca = pca.transform(scaled_data)
```

```python
scaled_data.shape # Telling us rows and cols
```

```python
x_pca.shape # reduced cols to 2
```

```python
#making visualization
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0], x_pca[:,1], c=cancer['target'], cmap="plasma")
plt.xlabel('First Principal component')
plt.ylabel('Second Principal component')
```

```python
# Showing the relationship of principal components and original cols
df_comp = pd.DataFrame(pca.components_, columns=cancer['feature_names'])
df_comp.head()
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(df_comp, cmap='plasma')
```
